# Taller: Factorización de Matrices en Ciencia de Datos

```python
import numpy as np
from scipy.linalg import lu, cholesky, qr, svd

# Matriz genérica (para LU y QR)
A = np.array([[2, 3, 1],
              [4, 7, 2],
              [6, 18, -1]], dtype=float)

# Matriz simétrica definida positiva (para Cholesky)
B = np.array([[25, 15, -5],
              [15, 18,  0],
              [-5,  0, 11]], dtype=float)
```

## Descomponer A=LU y verificar la reconstrucción.

```python
P, L, U = lu(A)
print("Matriz L:\n", L)
print("Matriz U:\n", U)

# Verificación
print("Reconstrucción:\n", P @ L @ U)
```

## Usar B=LL^T

```python
L_chol = cholesky(B, lower=True)
print("Matriz L (Cholesky):\n", L_chol)
print("Reconstrucción:\n", L_chol @ L_chol.T)
```

## Descomponer A = QR

```python
Q, R = qr(A)
print("Matriz Q:\n", Q)
print("Matriz R:\n", R)
print("Reconstrucción:\n", Q @ R)
```

## Descomponer y Analizar valores singulares

```python
U, s, Vt = svd(A)
print("Valores singulares:", s)
print("Reconstrucción:\n", U @ np.diag(s) @ Vt)
```
