# Compresión de imagen con SVD truncado

```python
import numpy as np
import matplotlib.pyplot as plt

# Cargar desde CSV
A = np.loadtxt("06_imagen_matriz.csv", delimiter=",")
```

## Descomposición SVD de la imagen

```python
import matplotlib.pyplot as plt
import numpy as np

U, s, Vt = np.linalg.svd(A, full_matrices=False)
```

## Función para reconstruir la imagen con k valores singulares

```python
def reconstruct_image(k):
    return U[:, :k] @ np.diag(s[:k]) @ Vt[:k, :]
```

## Reconstrucciones con distintos k

```python
ks = [5, 20, 50, 100]
plt.figure(figsize=(20, 4))

for i, k in enumerate(ks, start=2):
    plt.subplot(1, len(ks)+1, i)
    plt.imshow(reconstruct_image(k), cmap="gray")
    plt.title(f"k={k}")
    plt.axis("off")

plt.suptitle("Compresión de imágenes con SVD truncado", fontsize=14)
plt.show()
```
